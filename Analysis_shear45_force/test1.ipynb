```python
import pandas as pd
import glob
import os

# Define the folder path
folder_path = '/Users/komert/Documents/GitHub/Senior_Project_CU/shear45'

# Find all CSV files in the folder
csv_files = glob.glob(os.path.join(folder_path, '*.csv'))

# Read them into a list of DataFrames
dataframes = []
for file in csv_files:
    df = pd.read_csv(file)
    dataframes.append(df)
    print(f"Loaded: {os.path.basename(file)}")

# Optional: If all CSVs share the same columns, combine them into one massive DataFrame
# combined_df = pd.concat(dataframes, ignore_index=True)
```

```python
# Access individual dataframes from the list
for i, df in enumerate(dataframes):
    print(f"DataFrame {i}:")
    print(df.head())
```
